```python
# Question1 1. Write a program to input a string and display:
               * Number of vowels
               * Number of consonants
               * Number of spaces
```

```python
string = input("Enter a string: ")

vowels = 0
consonants = 0
spaces = 0

for ch in string:
    if ch.lower() in "aeiou":
        vowels += 1
    elif ch.isalpha():
        consonants += 1
    elif ch == " ":
        spaces += 1

print("Number of vowels =", vowels)
print("Number of consonants =", consonants)
print("Number of spaces =", spaces)
```

```python
# Question2 2. Write a program that takes a list of numbers and:
                * Prints the largest and smallest number.
                * Calculates the sum of all numbers.
```

```python
# Input a list of numbers separated by spaces
numbers = list(map(int, input("Enter numbers: ").split()))

largest = max(numbers)
smallest = min(numbers)
total = sum(numbers)

print("Largest number =", largest)
print("Smallest number =", smallest)
print("Sum of all numbers =", total)
```

```python
# Question3 Write a program to remove duplicates from a list.
```

```python
# Input a list of numbers separated by spaces
numbers = list(map(int, input("Enter numbers: ").split()))

# Remove duplicates
unique_numbers = list(set(numbers))

print("List after removing duplicates:", unique_numbers)
```

```python
# queston4 Write a program to reverse a string entered by the user
```

```python
# Input a string
string = input("Enter a string: ")

# Reverse the string
reversed_string = string[::-1]

print("Reversed string:", reversed_string)
```

```python
# Question5 Write a program to take names as a comma-separated string and display them as a sorted list
```

```python
# Input names as a comma-separated string
names = input("Enter names separated by commas: ")

# Convert string to list and remove extra spaces
name_list = [name.strip() for name in names.split(",")]

# Sort the list
name_list.sort()

print("Sorted list of names:")
print(name_list)
```

```python
# Question6 Write a program that takes a sentence and prints how many words it has.
```

```python
# Input a sentence
sentence = input("Enter a sentence: ")

# Split the sentence into words
words = sentence.split()

# Count the words
count = len(words)

print("Number of words:", count)
```
